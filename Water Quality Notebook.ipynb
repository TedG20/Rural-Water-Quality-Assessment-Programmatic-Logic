# Rural Water Quality Assessment & Programmatic Logic
This notebook translates the Kenya Water Service Regulatory Board (WASREB) guidelines and the E. coli / Risk-Of-Contamination (ROC) risk matrix into executable Python functions


```python
def evaluate_water_safety(coliforms_present, e_coli_present, ph_value, ammonia_value):
    """
    Evaluates whether a water source requires action based on general safety standards:
    - Coliforms: Must be absent (False)
    - E. coli: Must be absent (False)
    - pH: Must be between 6.5 and 8.5 inclusive
    - Ammonia: Must not exceed 0.5 mg/l
    """
    if coliforms_present or e_coli_present or ph_value < 6.5 or ph_value > 8.5 or ammonia_value > 0.5:
        return "Action required"
    else:
        return "No action required"

# --- Testing Part 1 ---
print("Test 1 (Safe Water Source):", evaluate_water_safety(False, False, 7.2, 0.2))
print("Test 2 (Contaminated with Coliforms):", evaluate_water_safety(True, False, 7.0, 0.1))
print("Test 3 (pH out of bounds):", evaluate_water_safety(False, False, 6.0, 0.1))
```

## Part 2: Assessing Action Priority (Class vs. ROC Matrix)
Next, we evaluate intervention priority by combining the **E. coli Classification (Classes A through E)** with the **Risk-Of-Contamination (ROC) score (0 to 9)**

```python
def assess_action_priority(col_class, roc_score):
    """
    Determines action priority tiers based on E. coli class and ROC score
    - No action
    - Low action priority
    - Higher action priority
    - Urgent action
    """
    # Normalize class input to uppercase
    col_class = col_class.upper()

    if col_class == 'A' and roc_score == 0:
        return "No action"
    elif (col_class == 'A' or col_class == 'B') and roc_score < 4:
        return "Low action priority"
    elif col_class in ['D', 'E'] or roc_score >= 7:
        return "Urgent action"
    else:
        return "Higher action priority"

# --- Testing Part 2 Matrix Cases ---
test_cases = [
    ('A', 0),  # Expected: No action
    ('A', 2),  # Expected: Low action priority
    ('B', 3),  # Expected: Low action priority
    ('C', 5),  # Expected: Higher action priority
    ('D', 1),  # Expected: Urgent action (class D/E rule)
    ('A', 8)   # Expected: Urgent action (ROC >= 7 rule)
]

for sample_class, sample_roc in test_cases:
    priority = assess_action_priority(sample_class, sample_roc)
    print(f"Class: {sample_class}, ROC: {sample_roc} --> Output: {priority}")
```

```python
# --- Sample #104 Data ---
sample_coliforms = True
sample_e_coli = False
sample_ph = 7.0
sample_ammonia = 0.3
sample_class = 'B'
sample_roc = 2

# 1. Run General Safety Evaluation
safety_result = evaluate_water_safety(sample_coliforms, sample_e_coli, sample_ph, sample_ammonia)

# 2. Run Action Priority Assessment
priority_result = assess_action_priority(sample_class, sample_roc)

# Print Final Report
print(f"--- Water Source Assessment Report ---")
print(f"Safety Status: {safety_result}")
print(f"Intervention Priority: {priority_result}")
```
